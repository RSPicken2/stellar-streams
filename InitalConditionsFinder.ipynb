```python
from galpy.potential import MWPotential2014
from galpy.orbit import Orbit
import matplotlib.pyplot as plt
import numpy as np

from astropy import units as u
from astropy.coordinates import SkyCoord, Galactocentric
```

```python
r = [-1.76, 11.37, 0.11]

Vz = 0.0
Vy = 2.4
Vxs = np.linspace(0, 225, 1000)


ics = [SkyCoord(
        x=r[0] * u.kpc,
        y=r[1] * u.kpc,
        z=r[2] * u.kpc,
        v_x = vx * (u.km/u.s),
        v_y = Vy * (u.km/u.s),
        v_z = Vz * (u.km/u.s),
        frame=Galactocentric()
    ) for vx in Vxs]
        
```

```python
o = Orbit(ics[205])
o.integrate(MWPotential2014)
print(o.e())
o.plot(d1="x", d2="y")
```

```python
orbits = [Orbit(sk) for sk in ics]
for o in orbits: 
    o.integrate(MWPotential2014)

es = np.asarray([o.e() for o in orbits])
```

```python
targets = np.linspace(0.2,0.8,7)
idxs = [np.argmin(np.abs(es - target)) for target in targets]
print(targets)
print(es[idxs])
print(Vxs[idxs].astype(int))
```

```python
plt.plot(Vxs, es)
plt.show()
```
